```python
urllib.request.urlretrieve("https://raw.githubusercontent.com/nshaud/ml_for_astro/main/stars.csv", "stars.csv")
import pandas as pd
df_stars = pd.read_csv("stars.csv")
```

```python
#name_list = ['temp', 'lum', 'radius', 'abs_mag', 'typ', 'color', 'spec_class']

temp = df_stars[df_stars.columns [0]]
lum = df_stars[df_stars.columns [1]]
radius = df_stars[df_stars.columns [2]]
abs_mag = df_stars[df_stars.columns [3]]
typ = df_stars[df_stars.columns [4]]
color = df_stars[df_stars.columns [5]]
spec_class = df_stars[df_stars.columns [6]]
```

```python
plt.scatter (temp, lum, c='black', s=10, alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.gca().invert_xaxis()

mean_temp = np.mean(temp)
mean_lum = np.mean(lum)
print (mean_temp, 'K', mean_lum, 'L_sun')

plt.plot ([min(temp), max(temp)], [mean_lum, mean_lum], ls='dashed')
plt.vlines (mean_temp, min(lum), max(lum), ls='dashed')
plt.xlim(0.9*min(temp), 1.1*max(temp))
plt.show()
```
