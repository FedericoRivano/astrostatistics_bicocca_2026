```python
import numpy as np 
from matplotlib import pyplot as plt
import sklearn as sk
from sklearn.manifold import Isomap
```

```python
loaddigits = sk.datasets.load_digits()
print(loaddigits.images.shape)
print(loaddigits.keys())
print(loaddigits.DESCR)
```

```python
X, _ = sk.datasets.load_digits(return_X_y=True)
X.shape
```

```python
embedding = Isomap(n_components=48)
X_transformed = embedding.fit_transform(X)
X_transformed.shape
```

```python
Test = X_transformed[np.random.randint(0, 1796)]
Test.shape
```

```python
plt.imshow(Test.reshape(8,6), cmap='binary')
```
