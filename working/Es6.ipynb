```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import sklearn.mixture as skm
```

```python
data = np.load ('/Users/federico.rivano/astrostatistics_bicocca_2026/solutions/formationchannels.npy')
```

```python
print (data.shape)
```

```python
plt.hist (data, bins = 1000)
plt.show()
```

```python
mode = skm.GaussianMixture(n_components=3)
mode.fit(data)
lL = mode.score_samples(data)
L = np.exp(lL)
print (lL)
print (L)
```

```python
x = np.linspace(data.min(), data.max(), len(data)).reshape(-1, 1)


plt.plot(x, L)
plt.hist(data, bins=50, density=True, alpha=0.5)
plt.show()
```
